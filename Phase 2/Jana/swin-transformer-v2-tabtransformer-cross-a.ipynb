```python
"""
╔══════════════════════════════════════════════════════════════════════════════╗
║     ISIC-DICM-17K — Transformer-Based Fusion  (v3 — XAI-Ready)             ║
║        Swin Transformer V2  ×  TabTransformer  +  Cross-Attention           ║
║   XAI Hooks: GradCAM++, SHAP, Integrated Gradients, Attention Rollout       ║
║                   Melanoma Classification — Colab/Kaggle GPU                ║
╚══════════════════════════════════════════════════════════════════════════════╝

Architecture:
  Image  → Swin V2 Base (pretrained) → patch tokens (B, N, D)
  Meta   → TabTransformer            → feature tokens (B, 23, D)
  Fusion → Bidirectional Cross-Attention × 2 layers
         → CLS tokens concat → classifier head → 2-class output

XAI Methods (run in separate XAI cell):
  1. GradCAM++             — spatial heatmap on final Swin feature map
  2. SHAP (DeepExplainer)  — feature importance for tabular metadata
  3. Integrated Gradients  — multimodal attribution (image + metadata)
  4. Attention Rollout     — propagated attention across all Transformer layers

Install:
    !pip install -q timm shap captum
"""

# ─── 0. Imports ───────────────────────────────────────────────────────────────
import os
import gc
import re
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from PIL import Image
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms
import timm

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import (
    roc_auc_score, classification_report,
    confusion_matrix, roc_curve, f1_score
)


# ─── 1. Configuration ─────────────────────────────────────────────────────────
class CFG:
    # ── Paths (UPDATE THESE) ──────────────────────────────────────────────────
    CSV_PATH    = "/kaggle/input/datasets/janaalnajjar/final-csv/final_preprocessed_data.csv"
    IMAGES_DIR  = "/kaggle/input/datasets/ahmedmohsen2005/xai-preprocessed-skin-lesion-17k-dataset/final_processed_images"
    OUTPUT_DIR  = "/kaggle/working/SwinV2_output"

    # ── Column names ──────────────────────────────────────────────────────────
    IMG_COL     = "image"
    TARGET_COL  = "class"

    DROP_COLS   = [
        "image", "isic_id", "patient_id", "class",
        "diagnosis_confirm_type_histopathology",   # LEAKAGE  r=0.53
        "anatom_site_general_oral/genital",         # DEAD     100% zeros
        "age_group_0-20", "age_group_21-40",        # REDUNDANT with age_scaled
        "age_group_41-60", "age_group_61-85",
        "year",                                     # TEMPORAL BIAS r=0.33
    ]
    # 30 raw features → 23 clean clinical features

    # ── Image ─────────────────────────────────────────────────────────────────
    IMG_SIZE    = 256
    MEAN        = [0.485, 0.456, 0.406]
    STD         = [0.229, 0.224, 0.225]

    # ── Transformer dims ──────────────────────────────────────────────────────
    D             = 128
    NUM_HEADS     = 8
    FF_DIM        = 256
    NUM_CA_LAYERS = 2
    DROPOUT_TF    = 0.1

    # ── TabTransformer ────────────────────────────────────────────────────────
    TAB_DEPTH   = 2
    TAB_HEADS   = 4

    # ── Training ──────────────────────────────────────────────────────────────
    SEED          = 42
    EPOCHS        = 40
    BATCH_SIZE    = 8
    GRAD_ACCUM    = 2
    NUM_WORKERS   = 2
    PATIENCE      = 10
    LABEL_SMOOTH  = 0.05

    # ── Optimizer ─────────────────────────────────────────────────────────────
    LR_SWIN       = 3e-5
    LR_HEAD       = 2e-4
    WEIGHT_DECAY  = 1e-4
    WARMUP_EPOCHS = 3

    # ── EMA early stopping ────────────────────────────────────────────────────
    EMA_ALPHA     = 0.3

    # ── K-Fold ────────────────────────────────────────────────────────────────
    USE_KFOLD   = False
    N_FOLDS     = 5

    # ── Swin V2 backbone ──────────────────────────────────────────────────────
    SWIN_MODEL  = "swinv2_base_window12to16_192to256_22kft1k"
    # SWIN_MODEL = "swinv2_small_window16_256"   # lighter option for T4

    DEVICE      = "cuda" if torch.cuda.is_available() else "cpu"


os.makedirs(CFG.OUTPUT_DIR, exist_ok=True)
torch.manual_seed(CFG.SEED)
np.random.seed(CFG.SEED)
print(f"Device : {CFG.DEVICE}")
print(f"GPU    : {torch.cuda.get_device_name(0) if CFG.DEVICE == 'cuda' else 'N/A'}")


# ─── 2. Feature Validator ─────────────────────────────────────────────────────
def validate_features(df, meta_cols):
    print("\n── Feature validation ──")
    issues = []
    for c in meta_cols:
        if df[c].std() == 0:
            issues.append(f"DEAD COLUMN (zero variance): {c}")
    for c in meta_cols:
        r = abs(df[c].corr(df[CFG.TARGET_COL]))
        if r > 0.45:
            issues.append(f"LEAKAGE RISK (r={r:.3f} with class): {c}")
    if issues:
        for i in issues:
            print(f"  {i}")
        raise ValueError("Fix issues above — add column names to CFG.DROP_COLS.")
    sparse = [(c, (df[c] == 0).mean()) for c in meta_cols
              if (df[c] == 0).mean() > 0.99]
    if sparse:
        print("  Sparse features (>99% zeros) — keeping as missing-data signal:")
        for c, pct in sparse:
            print(f"    {c} ({pct*100:.1f}% zeros)")
    print(f"  Passed — {len(meta_cols)} clean features ready for training.\n")


# ─── 3. EMA Tracker ───────────────────────────────────────────────────────────
class EMAMetric:
    def __init__(self, alpha=0.3):
        self.alpha = alpha
        self.value = None

    def update(self, v):
        self.value = v if self.value is None \
            else self.alpha * v + (1 - self.alpha) * self.value
        return self.value


# ─── 4. Threshold Optimizer ───────────────────────────────────────────────────
def find_optimal_threshold(labels, probs):
    fpr, tpr, thresholds = roc_curve(labels, probs)
    j        = tpr - fpr
    best_idx = int(np.argmax(j))
    return float(thresholds[best_idx]), float(j[best_idx])


# ─── 5. Dataset ───────────────────────────────────────────────────────────────
class ISICDataset(Dataset):
    def __init__(self, df, images_dir, augment=False):
        self.df         = df.reset_index(drop=True)
        self.images_dir = images_dir
        self.meta_cols  = [c for c in df.columns if c not in CFG.DROP_COLS]
        self.meta_dim   = len(self.meta_cols)

        if augment:
            self.transform = transforms.Compose([
                transforms.Resize((CFG.IMG_SIZE, CFG.IMG_SIZE)),
                transforms.RandomHorizontalFlip(p=0.5),
                transforms.RandomVerticalFlip(p=0.5),
                transforms.RandomRotation(degrees=20),
                transforms.ColorJitter(
                    brightness=0.15, contrast=0.15,
                    saturation=0.1,  hue=0.03
                ),
                transforms.RandomAffine(
                    degrees=0, translate=(0.05, 0.05), scale=(0.95, 1.05)
                ),
                transforms.ToTensor(),
                transforms.Normalize(CFG.MEAN, CFG.STD),
            ])
        else:
            self.transform = transforms.Compose([
                transforms.Resize((CFG.IMG_SIZE, CFG.IMG_SIZE)),
                transforms.ToTensor(),
                transforms.Normalize(CFG.MEAN, CFG.STD),
            ])

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
            row = self.df.iloc[idx]
            
            # ─── FIX: Standardize Image Name (Regex) ───────────────────────────
            # This extracts exactly 7 digits from whatever is in the CSV cell
            # e.g., 'ISIC2019_0013702_mel.jpg' -> 'ISIC_0013702.jpg'
            raw_name = str(row[CFG.IMG_COL])
            match = re.search(r'(\d{7})', raw_name)
            img_name = f"ISIC_{match.group(1)}.jpg" if match else raw_name
            # ───────────────────────────────────────────────────────────────────
    
            img_path = os.path.join(self.images_dir, img_name)
            
            if not os.path.exists(img_path):
                raise FileNotFoundError(f"CRITICAL ERROR: Missing image {img_name} at {img_path}. "
                                        f"Check your IMAGES_DIR or Regex logic.")
    
            image = Image.open(img_path).convert("RGB")
            image = self.transform(image)
            
            meta = torch.tensor(
                row[self.meta_cols].values.astype(np.float32),
                dtype=torch.float32
            )
            label = torch.tensor(int(row[CFG.TARGET_COL]), dtype=torch.long)
            
            return image, meta, label


def make_weighted_sampler(labels, epoch=0):
    class_counts  = np.bincount(labels)
    class_weights = 1.0 / class_counts
    sample_w      = class_weights[labels]
    return WeightedRandomSampler(
        weights=torch.from_numpy(sample_w).float(),
        num_samples=len(sample_w),
        replacement=True,
        generator=torch.Generator().manual_seed(CFG.SEED + epoch)
    )


# ─── 6. Model Components ──────────────────────────────────────────────────────

# ── 6a. Swin V2 Image Encoder ─────────────────────────────────────────────────
class SwinV2ImageEncoder(nn.Module):
    """
    Swin V2 Base — returns patch token sequence (B, N+1, D).
    XAI hooks:
      self.feature_map   — raw (B, H, W, C) feature map BEFORE projection
                           → used by GradCAM++ and Integrated Gradients
      self.grad_map      — gradient w.r.t. feature_map (set by GradCAM++ hook)
    """
    def __init__(self, out_dim=128, dropout=0.1):
        super().__init__()
        self.backbone = timm.create_model(
            CFG.SWIN_MODEL,
            pretrained=True,
            num_classes=0,
            global_pool="",
        )
        self._freeze_stages(freeze_up_to=2)

        with torch.no_grad():
            dummy = torch.zeros(1, 3, CFG.IMG_SIZE, CFG.IMG_SIZE)
            feats = self.backbone.forward_features(dummy)
            self.spatial_flatten = feats.dim() == 4
            self.feat_dim        = feats.shape[-1]

        self.proj = nn.Sequential(
            nn.LayerNorm(self.feat_dim),
            nn.Linear(self.feat_dim, out_dim),
            nn.Dropout(dropout),
        )

        self.cls_token = nn.Parameter(torch.zeros(1, 1, out_dim))
        nn.init.trunc_normal_(self.cls_token, std=0.02)

        # XAI state
        self.feature_map  = None   # (B, H, W, C) — set during forward
        self.grad_map     = None   # (B, H, W, C) — set by GradCAM++ backward hook
        self._hook_handle = None

    def _freeze_stages(self, freeze_up_to=2):
        for name, param in self.backbone.named_parameters():
            freeze = "patch_embed" in name or "conv_first" in name
            for i in range(freeze_up_to):
                if f"layers.{i}" in name or f"stages.{i}" in name:
                    freeze = True
            if freeze:
                param.requires_grad = False

    def register_gradcam_hook(self):
        """
        Registers a backward hook on the last Swin stage output.
        Stores gradients in self.grad_map for GradCAM++ computation.
        Call once before running GradCAM++; call remove_gradcam_hook() after.
        """
        def _save_grad(grad):
            self.grad_map = grad.detach()

        def _hook_fn(module, input, output):
            # output is (B, H, W, C) for Swin; register hook on the tensor
            if isinstance(output, torch.Tensor):
                output.register_hook(_save_grad)
            return output

        # Hook onto the last Swin stage (layers[-1] or stages[-1])
        if hasattr(self.backbone, "layers"):
            target = self.backbone.layers[-1]
        elif hasattr(self.backbone, "stages"):
            target = self.backbone.stages[-1]
        else:
            raise AttributeError("Cannot find Swin stages for GradCAM++ hook.")

        self._hook_handle = target.register_forward_hook(_hook_fn)

    def remove_gradcam_hook(self):
        if self._hook_handle is not None:
            self._hook_handle.remove()
            self._hook_handle = None

    def forward(self, x):
        feats = self.backbone.forward_features(x)
        if self.spatial_flatten and feats.dim() == 4:
            B, H, W, C = feats.shape
            self.feature_map = feats                     # store (B,H,W,C) for XAI
            feats = feats.view(B, H * W, C)
        else:
            self.feature_map = feats
        tokens = self.proj(feats)
        cls    = self.cls_token.expand(tokens.size(0), -1, -1)
        return torch.cat([cls, tokens], dim=1)


# ── 6b. TabTransformer Metadata Encoder ───────────────────────────────────────
class TabTransformerEncoder(nn.Module):
    """
    23 clinical features → (B, 24, D).
    All TransformerEncoder layer attention weights are stored in
    self.layer_attn_weights (list of tensors) for Attention Rollout.
    """
    def __init__(self, num_features, embed_dim=128, depth=2,
                 num_heads=4, ff_dim=256, dropout=0.1):
        super().__init__()
        self.num_features = num_features
        self.embed_dim    = embed_dim
        self.num_heads    = num_heads

        self.feature_embeddings = nn.ModuleList([
            nn.Sequential(nn.Linear(1, embed_dim), nn.LayerNorm(embed_dim))
            for _ in range(num_features)
        ])
        self.pos_emb = nn.Parameter(torch.zeros(1, num_features, embed_dim))
        nn.init.trunc_normal_(self.pos_emb, std=0.02)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, embed_dim))
        nn.init.trunc_normal_(self.cls_token, std=0.02)

        # Build individual attention layers so we can extract weights
        self.attn_layers = nn.ModuleList([
            nn.MultiheadAttention(
                embed_dim=embed_dim, num_heads=num_heads,
                dropout=dropout, batch_first=True
            )
            for _ in range(depth)
        ])
        self.ffn_layers = nn.ModuleList([
            nn.Sequential(
                nn.LayerNorm(embed_dim),
                nn.Linear(embed_dim, ff_dim),
                nn.GELU(),
                nn.Dropout(dropout),
                nn.Linear(ff_dim, embed_dim),
                nn.Dropout(dropout),
            )
            for _ in range(depth)
        ])
        self.layer_norms = nn.ModuleList([
            nn.LayerNorm(embed_dim) for _ in range(depth)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        # XAI: stores per-layer attention weights (B, heads, seq, seq)
        self.layer_attn_weights = []

    def forward(self, x):
        B = x.size(0)
        tokens = torch.stack([
            self.feature_embeddings[i](x[:, i:i+1])
            for i in range(self.num_features)
        ], dim=1) + self.pos_emb
        cls    = self.cls_token.expand(B, -1, -1)
        tokens = torch.cat([cls, tokens], dim=1)

        # Clear previous attention weights
        self.layer_attn_weights = []

        for attn, ffn, ln in zip(self.attn_layers, self.ffn_layers, self.layer_norms):
            normed = ln(tokens)
            attn_out, attn_w = attn(
                normed, normed, normed,
                need_weights=True, average_attn_weights=False
            )
            tokens = tokens + attn_out
            tokens = tokens + ffn(tokens)
            self.layer_attn_weights.append(attn_w.detach())  # (B, heads, seq, seq)

        return self.norm(tokens)


# ── 6c. Cross-Attention Fusion Block ──────────────────────────────────────────
class CrossAttentionBlock(nn.Module):
    """
    Bidirectional cross-attention.
    Stores attention weights for XAI and Attention Rollout.
    """
    def __init__(self, dim, num_heads, ff_dim, dropout=0.1):
        super().__init__()
        self.img_to_meta = nn.MultiheadAttention(
            embed_dim=dim, num_heads=num_heads,
            dropout=dropout, batch_first=True
        )
        self.meta_to_img = nn.MultiheadAttention(
            embed_dim=dim, num_heads=num_heads,
            dropout=dropout, batch_first=True
        )
        self.ffn_img = nn.Sequential(
            nn.LayerNorm(dim), nn.Linear(dim, ff_dim),
            nn.GELU(), nn.Dropout(dropout),
            nn.Linear(ff_dim, dim), nn.Dropout(dropout),
        )
        self.ffn_meta = nn.Sequential(
            nn.LayerNorm(dim), nn.Linear(dim, ff_dim),
            nn.GELU(), nn.Dropout(dropout),
            nn.Linear(ff_dim, dim), nn.Dropout(dropout),
        )
        self.norm_img  = nn.LayerNorm(dim)
        self.norm_meta = nn.LayerNorm(dim)

    def forward(self, img_tokens, meta_tokens):
        img_ca, attn_i2m = self.img_to_meta(
            query=self.norm_img(img_tokens),
            key=meta_tokens, value=meta_tokens,
            need_weights=True, average_attn_weights=False
        )
        img_tokens  = img_tokens  + img_ca
        img_tokens  = img_tokens  + self.ffn_img(img_tokens)

        meta_ca, attn_m2i = self.meta_to_img(
            query=self.norm_meta(meta_tokens),
            key=img_tokens, value=img_tokens,
            need_weights=True, average_attn_weights=False
        )
        meta_tokens = meta_tokens + meta_ca
        meta_tokens = meta_tokens + self.ffn_meta(meta_tokens)

        return img_tokens, meta_tokens, attn_i2m, attn_m2i


# ── 6d. Full Multimodal Transformer Model ─────────────────────────────────────
class TransformerFusionModel(nn.Module):
    """
    Full pipeline with XAI hooks:
      - self._attn_img, self._attn_meta  : last-layer cross-attention weights
      - self._all_ca_attn_img            : all CA layer attn_i2m weights (rollout)
      - self._all_ca_attn_meta           : all CA layer attn_m2i weights (rollout)
      - self.img_encoder.feature_map     : raw Swin feature map (GradCAM++)
      - self.meta_encoder.layer_attn_weights : TabTransformer self-attn (rollout)
    """
    def __init__(self, meta_dim):
        super().__init__()
        D = CFG.D

        self.img_encoder  = SwinV2ImageEncoder(out_dim=D, dropout=CFG.DROPOUT_TF)
        self.meta_encoder = TabTransformerEncoder(
            num_features=meta_dim, embed_dim=D,
            depth=CFG.TAB_DEPTH, num_heads=CFG.TAB_HEADS,
            ff_dim=CFG.FF_DIM, dropout=CFG.DROPOUT_TF
        )
        self.ca_layers = nn.ModuleList([
            CrossAttentionBlock(
                dim=D, num_heads=CFG.NUM_HEADS,
                ff_dim=CFG.FF_DIM, dropout=CFG.DROPOUT_TF
            )
            for _ in range(CFG.NUM_CA_LAYERS)
        ])

        self.classifier = nn.Sequential(
            nn.LayerNorm(D * 2),
            nn.Linear(D * 2, 256),
            nn.GELU(), nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.GELU(), nn.Dropout(0.2),
            nn.Linear(128, 2),
        )

        # XAI state
        self._attn_img       = None
        self._attn_meta      = None
        self._all_ca_attn_img  = []
        self._all_ca_attn_meta = []

    def forward(self, image, meta):
        img_tokens  = self.img_encoder(image)
        meta_tokens = self.meta_encoder(meta)

        self._all_ca_attn_img  = []
        self._all_ca_attn_meta = []

        for ca in self.ca_layers:
            img_tokens, meta_tokens, a_i, a_m = ca(img_tokens, meta_tokens)
            self._all_ca_attn_img.append(a_i.detach())
            self._all_ca_attn_meta.append(a_m.detach())

        self._attn_img  = self._all_ca_attn_img[-1]
        self._attn_meta = self._all_ca_attn_meta[-1]

        img_cls  = img_tokens[:, 0, :]
        meta_cls = meta_tokens[:, 0, :]
        fused    = torch.cat([img_cls, meta_cls], dim=1)
        return self.classifier(fused)

    def get_attention_weights(self):
        return self._attn_img, self._attn_meta

    def get_all_ca_attention_weights(self):
        """Returns list of (attn_i2m, attn_m2i) per CA layer — for Rollout."""
        return list(zip(self._all_ca_attn_img, self._all_ca_attn_meta))

    def get_tab_attn_weights(self):
        """Returns list of per-layer self-attn from TabTransformer — for Rollout."""
        return self.meta_encoder.layer_attn_weights


# ─── 7. Loss ──────────────────────────────────────────────────────────────────
class LabelSmoothingCrossEntropy(nn.Module):
    def __init__(self, smoothing=0.05):
        super().__init__()
        self.smoothing = smoothing

    def forward(self, logits, targets):
        n  = logits.size(1)
        lp = F.log_softmax(logits, dim=1)
        with torch.no_grad():
            s = torch.full_like(lp, self.smoothing / (n - 1))
            s.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)
        return -(s * lp).sum(dim=1).mean()


# ─── 8. LR Scheduler with Warmup ─────────────────────────────────────────────
class WarmupCosineScheduler:
    def __init__(self, optimizer, warmup_epochs, total_epochs, min_lr=1e-7):
        self.opt           = optimizer
        self.warmup_epochs = warmup_epochs
        self.total_epochs  = total_epochs
        self.min_lr        = min_lr
        self.base_lrs      = [pg["lr"] for pg in optimizer.param_groups]

    def step(self, epoch):
        if epoch < self.warmup_epochs:
            scale = max(0.01, (epoch + 1) / self.warmup_epochs)
        else:
            progress = (epoch - self.warmup_epochs) / \
                       max(1, self.total_epochs - self.warmup_epochs)
            scale = max(
                0.5 * (1 + math.cos(math.pi * progress)),
                self.min_lr / max(self.base_lrs)
            )
        for pg, base in zip(self.opt.param_groups, self.base_lrs):
            pg["lr"] = base * scale

    def get_lr(self):
        return [pg["lr"] for pg in self.opt.param_groups]


# ─── 9. Train / Eval Loops ────────────────────────────────────────────────────
def train_one_epoch(model, loader, optimizer, criterion, scaler_amp, device):
    model.train()
    total_loss = 0
    all_labels, all_probs = [], []
    optimizer.zero_grad()

    for step, (images, metas, labels) in enumerate(
            tqdm(loader, desc="  Train", leave=False)):

        images = images.to(device, non_blocking=True)
        metas  = metas.to(device,  non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.amp.autocast("cuda", enabled=(device == "cuda")):
            logits = model(images, metas)
            loss   = criterion(logits, labels) / CFG.GRAD_ACCUM

        scaler_amp.scale(loss).backward()

        if (step + 1) % CFG.GRAD_ACCUM == 0:
            scaler_amp.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=0.5)
            scaler_amp.step(optimizer)
            scaler_amp.update()
            optimizer.zero_grad()

        total_loss += loss.item() * CFG.GRAD_ACCUM * images.size(0)
        probs = torch.softmax(logits, dim=1)[:, 1].detach().float().cpu().numpy()
        all_probs.extend(probs)
        all_labels.extend(labels.cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    auc      = roc_auc_score(all_labels, all_probs)
    return avg_loss, auc


@torch.no_grad()
def evaluate(model, loader, criterion, device, threshold=0.5):
    model.eval()
    total_loss = 0
    all_labels, all_probs, all_preds = [], [], []

    for images, metas, labels in tqdm(loader, desc="  Val  ", leave=False):
        images = images.to(device, non_blocking=True)
        metas  = metas.to(device,  non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.amp.autocast("cuda", enabled=(device == "cuda")):
            logits = model(images, metas)
            loss   = criterion(logits, labels)

        total_loss += loss.item() * images.size(0)
        probs  = torch.softmax(logits, dim=1)[:, 1].float().cpu().numpy()
        preds  = (probs >= threshold).astype(int)
        all_probs.extend(probs)
        all_preds.extend(preds)
        all_labels.extend(labels.cpu().numpy())

    avg_loss  = total_loss / len(loader.dataset)
    labels_np = np.array(all_labels)
    probs_np  = np.array(all_probs)
    preds_np  = np.array(all_preds)
    auc       = roc_auc_score(labels_np, probs_np)
    f1        = f1_score(labels_np, preds_np, average="macro")
    return avg_loss, auc, f1, labels_np, preds_np, probs_np


# ─── 10. Standard Plots ───────────────────────────────────────────────────────
def plot_training_curves(history, fold=None):
    tag = f"_fold{fold}" if fold else ""
    fig, axes = plt.subplots(1, 4, figsize=(22, 5))

    axes[0].plot(history["train_loss"], label="Train", marker="o", ms=3)
    axes[0].plot(history["val_loss"],   label="Val",   marker="o", ms=3)
    axes[0].set_title("Loss"); axes[0].legend(); axes[0].grid(True)

    axes[1].plot(history["train_auc"],   label="Train raw",  ms=3, alpha=0.4)
    axes[1].plot(history["val_auc"],     label="Val raw",    ms=3, alpha=0.4)
    axes[1].plot(history["val_auc_ema"], label="Val EMA", lw=2)
    axes[1].set_title("AUC-ROC"); axes[1].legend(); axes[1].grid(True)

    axes[2].plot(history["val_f1"], label="Val F1", color="green", ms=3)
    axes[2].set_title("Val F1 macro"); axes[2].legend(); axes[2].grid(True)

    axes[3].plot(history["threshold"], label="Youden-J threshold", color="purple", ms=3)
    axes[3].axhline(0.5, color="gray", linestyle="--", alpha=0.5, label="Default 0.5")
    axes[3].set_title("Optimal threshold"); axes[3].legend(); axes[3].grid(True)

    plt.suptitle(
        f"Swin V2 × TabTransformer v3{f' | Fold {fold}' if fold else ''}",
        fontsize=13, fontweight="bold"
    )
    plt.tight_layout()
    path = os.path.join(CFG.OUTPUT_DIR, f"training_curves{tag}.png")
    plt.savefig(path, dpi=150); plt.show(); plt.close()
    print(f"Saved: {path}")


def plot_confusion_matrix(labels, preds, threshold, fold=None):
    tag = f"_fold{fold}" if fold else ""
    cm  = confusion_matrix(labels, preds)
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Purples",
                xticklabels=["Non-Melanoma", "Melanoma"],
                yticklabels=["Non-Melanoma", "Melanoma"], ax=ax)
    ax.set_ylabel("True"); ax.set_xlabel("Predicted")
    ax.set_title(f"Confusion Matrix (threshold={threshold:.3f})")
    plt.tight_layout()
    path = os.path.join(CFG.OUTPUT_DIR, f"confusion_matrix{tag}.png")
    plt.savefig(path, dpi=150); plt.show(); plt.close()


def plot_roc_curve(labels, probs, auc_score, threshold, fold=None):
    tag = f"_fold{fold}" if fold else ""
    fpr, tpr, _ = roc_curve(labels, probs)
    fig, ax = plt.subplots(figsize=(6, 5))
    ax.plot(fpr, tpr, lw=2, color="purple", label=f"AUC = {auc_score:.4f}")
    ax.plot([0, 1], [0, 1], "k--")
    ax.set_xlabel("FPR"); ax.set_ylabel("TPR")
    ax.set_title(f"ROC Curve (optimal threshold={threshold:.3f})")
    ax.legend(); ax.grid(True)
    plt.tight_layout()
    path = os.path.join(CFG.OUTPUT_DIR, f"roc_curve{tag}.png")
    plt.savefig(path, dpi=150); plt.show(); plt.close()


def plot_calibration_curve(labels, probs, fold=None):
    tag    = f"_fold{fold}" if fold else ""
    n_bins = 10
    bins   = np.linspace(0, 1, n_bins + 1)
    centers, fracs = [], []
    for i in range(n_bins):
        mask = (probs >= bins[i]) & (probs < bins[i+1])
        if mask.sum() > 0:
            centers.append(probs[mask].mean())
            fracs.append(labels[mask].mean())
    fig, ax = plt.subplots(figsize=(6, 5))
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.plot(centers, fracs, marker="o", lw=2, color="purple", label="Model")
    ax.set_xlabel("Mean predicted P(Melanoma)")
    ax.set_ylabel("True melanoma fraction")
    ax.set_title("Calibration curve")
    ax.legend(); ax.grid(True)
    plt.tight_layout()
    path = os.path.join(CFG.OUTPUT_DIR, f"calibration{tag}.png")
    plt.savefig(path, dpi=150); plt.show(); plt.close()


# ── Built-in cross-attention visualization (lightweight, always runs) ─────────
def plot_cross_attention_heatmap(attn_weights, meta_cols,
                                  sample_idx=0, save_path=None):
    """Image CLS → clinical features attention bar chart."""
    attn    = attn_weights[sample_idx].mean(dim=0)
    cls_row = attn[0, 1:].cpu().numpy()
    n       = min(len(meta_cols), len(cls_row))
    vals    = cls_row[:n]
    cols    = meta_cols[:n]
    colors  = plt.cm.RdYlGn(vals / (vals.max() + 1e-8))
    fig, ax = plt.subplots(figsize=(max(14, n // 2), 3))
    ax.bar(range(n), vals, color=colors)
    ax.set_xticks(range(n))
    ax.set_xticklabels(cols, rotation=45, ha="right", fontsize=8)
    ax.set_ylabel("Attention weight")
    ax.set_title(
        "Cross-attention: Image CLS → Clinical features\n"
        "(Which clinical features guided the visual decision?)"
    )
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()


def plot_spatial_attention(attn_meta, meta_cols, raw_np,
                            sample_idx=0, save_path=None):
    """Clinical CLS → image patches — overlaid spatial heatmap."""
    meta_cls_attn = attn_meta[sample_idx].mean(dim=0)[0, 1:].cpu().numpy()
    n_patches     = len(meta_cls_attn)
    grid_size     = int(np.sqrt(n_patches))

    if grid_size * grid_size == n_patches:
        spatial = meta_cls_attn.reshape(grid_size, grid_size)
    else:
        pad = np.zeros(grid_size * grid_size)
        pad[:n_patches] = meta_cls_attn
        spatial = pad.reshape(grid_size, grid_size)

    spatial_up = np.array(
        Image.fromarray(
            (spatial / (spatial.max() + 1e-8) * 255).astype(np.uint8)
        ).resize((CFG.IMG_SIZE, CFG.IMG_SIZE), Image.BILINEAR)
    ) / 255.0

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(raw_np); axes[0].set_title("Original image"); axes[0].axis("off")
    im = axes[1].imshow(spatial, cmap="hot"); axes[1].set_title("Patch attention map")
    axes[1].axis("off"); plt.colorbar(im, ax=axes[1], fraction=0.046)
    axes[2].imshow(raw_np); axes[2].imshow(spatial_up, cmap="hot", alpha=0.5)
    axes[2].set_title("Clinical → Visual attention overlay"); axes[2].axis("off")
    plt.suptitle("Where did clinical metadata focus in the image?",
                 fontweight="bold")
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()


# ─── 11. Core Training Function ───────────────────────────────────────────────
def run_training(train_df, val_df, meta_cols, fold=None):
    tag = f"fold{fold}" if fold else "model"
    print(f"\n{'='*65}")
    print(f"  Swin V2 × TabTransformer — Cross-Attention Fusion  v3")
    print(f"  {'Fold ' + str(fold) if fold else 'Single Split'}")
    print(f"  Train: {len(train_df):,}  |  Val: {len(val_df):,}")
    print(f"  Meta features: {len(meta_cols)}")
    print(f"{'='*65}")

    train_ds = ISICDataset(train_df, CFG.IMAGES_DIR, augment=True)
    val_ds   = ISICDataset(val_df,   CFG.IMAGES_DIR, augment=False)

    model   = TransformerFusionModel(meta_dim=len(meta_cols)).to(CFG.DEVICE)
    total_p = sum(p.numel() for p in model.parameters())
    train_p = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"\n  Total params    : {total_p:,}")
    print(f"  Trainable params: {train_p:,}")

    swin_params = [p for n, p in model.named_parameters()
                   if "img_encoder" in n and p.requires_grad]
    head_params = [p for n, p in model.named_parameters()
                   if "img_encoder" not in n and p.requires_grad]

    optimizer  = optim.AdamW([
        {"params": swin_params, "lr": CFG.LR_SWIN},
        {"params": head_params, "lr": CFG.LR_HEAD},
    ], weight_decay=CFG.WEIGHT_DECAY)

    scheduler  = WarmupCosineScheduler(optimizer, CFG.WARMUP_EPOCHS, CFG.EPOCHS)
    criterion  = LabelSmoothingCrossEntropy(CFG.LABEL_SMOOTH)
    scaler_amp = torch.amp.GradScaler("cuda", enabled=(CFG.DEVICE == "cuda"))

    ema_auc    = EMAMetric(alpha=CFG.EMA_ALPHA)
    best_ema   = 0.0
    patience_c = 0
    best_thr   = 0.5
    ckpt_path  = os.path.join(CFG.OUTPUT_DIR, f"best_{tag}.pth")

    history = {
        "train_loss": [], "val_loss": [],
        "train_auc":  [], "val_auc": [], "val_auc_ema": [],
        "val_f1":     [], "threshold": []
    }

    for epoch in range(1, CFG.EPOCHS + 1):
        sampler = make_weighted_sampler(
            train_df[CFG.TARGET_COL].values, epoch=epoch
        )
        train_loader = DataLoader(
            train_ds, batch_size=CFG.BATCH_SIZE, sampler=sampler,
            num_workers=CFG.NUM_WORKERS, pin_memory=True
        )
        val_loader = DataLoader(
            val_ds, batch_size=CFG.BATCH_SIZE, shuffle=False,
            num_workers=CFG.NUM_WORKERS, pin_memory=True
        )

        lrs = scheduler.get_lr()
        print(f"\n  Epoch {epoch:02d}/{CFG.EPOCHS}  "
              f"| LR swin={lrs[0]:.1e}  head={lrs[1]:.1e}")

        tr_loss, tr_auc = train_one_epoch(
            model, train_loader, optimizer, criterion, scaler_amp, CFG.DEVICE
        )

        vl_loss, vl_auc, _, vl_labels, _, vl_probs = evaluate(
            model, val_loader, criterion, CFG.DEVICE, threshold=0.5
        )
        opt_thr, youden_j = find_optimal_threshold(vl_labels, vl_probs)
        vl_preds_opt      = (vl_probs >= opt_thr).astype(int)
        vl_f1_opt         = f1_score(vl_labels, vl_preds_opt, average="macro")

        smooth_auc = ema_auc.update(vl_auc)
        scheduler.step(epoch)

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(vl_loss)
        history["train_auc"].append(tr_auc)
        history["val_auc"].append(vl_auc)
        history["val_auc_ema"].append(smooth_auc)
        history["val_f1"].append(vl_f1_opt)
        history["threshold"].append(opt_thr)

        print(f"  Train → Loss: {tr_loss:.4f}  AUC: {tr_auc:.4f}")
        print(f"  Val   → Loss: {vl_loss:.4f}  AUC: {vl_auc:.4f}  "
              f"EMA: {smooth_auc:.4f}  F1: {vl_f1_opt:.4f}  "
              f"thresh: {opt_thr:.3f}  J: {youden_j:.3f}")

        if smooth_auc > best_ema:
            best_ema   = smooth_auc
            best_thr   = opt_thr
            patience_c = 0
            torch.save({
                "epoch":            epoch,
                "model_state_dict": model.state_dict(),
                "val_auc_raw":      vl_auc,
                "val_auc_ema":      smooth_auc,
                "val_f1":           vl_f1_opt,
                "threshold":        opt_thr,
                "meta_cols":        meta_cols,
                "meta_dim":         len(meta_cols),
            }, ckpt_path)
            print(f"  ✅ Saved  EMA-AUC: {best_ema:.4f}  threshold: {opt_thr:.3f}")
        else:
            patience_c += 1
            print(f"  ⏳ No improvement ({patience_c}/{CFG.PATIENCE})")
            if patience_c >= CFG.PATIENCE:
                print("  🛑 Early stopping.")
                break

        if CFG.DEVICE == "cuda":
            torch.cuda.empty_cache(); gc.collect()

    # ── Final evaluation ──────────────────────────────────────────────────────
    print(f"\n{'─'*65}")
    print(f"  FINAL RESULTS  (EMA-AUC: {best_ema:.4f}  threshold: {best_thr:.3f})")
    print(f"{'─'*65}")

    ckpt = torch.load(ckpt_path, map_location=CFG.DEVICE, weights_only=False)
    model.load_state_dict(ckpt["model_state_dict"])
    saved_thr = ckpt["threshold"]

    _, final_auc, _, final_labels, _, final_probs = evaluate(
        model, val_loader, criterion, CFG.DEVICE, threshold=0.5
    )
    final_preds = (final_probs >= saved_thr).astype(int)
    final_f1    = f1_score(final_labels, final_preds, average="macro")

    print(f"\n  AUC-ROC : {final_auc:.4f}")
    print(f"  F1 Macro (@ threshold {saved_thr:.3f}): {final_f1:.4f}")
    print(f"\n  Classification Report (@ optimal threshold {saved_thr:.3f}):")
    print(classification_report(
        final_labels, final_preds,
        target_names=["Non-Melanoma", "Melanoma"], digits=4
    ))

    # ── Built-in cross-attention XAI (always runs; advanced XAI in separate cell) ─
    print("\n  Generating built-in cross-attention visualizations...")
    model.eval()
    xai_loader = DataLoader(val_ds, batch_size=CFG.BATCH_SIZE,
                             shuffle=False, num_workers=0)
    sample_img, sample_meta, _ = next(iter(xai_loader))
    with torch.no_grad():
        with torch.amp.autocast("cuda", enabled=(CFG.DEVICE == "cuda")):
            _ = model(sample_img.to(CFG.DEVICE), sample_meta.to(CFG.DEVICE))

    attn_img, attn_meta = model.get_attention_weights()
    plot_cross_attention_heatmap(
        attn_img, meta_cols, sample_idx=0,
        save_path=os.path.join(
            CFG.OUTPUT_DIR,
            f"xai_clinical_attention{'_fold'+str(fold) if fold else ''}.png"
        )
    )

    plot_training_curves(history, fold=fold)
    plot_confusion_matrix(final_labels, final_preds, saved_thr, fold=fold)
    plot_roc_curve(final_labels, final_probs, final_auc, saved_thr, fold=fold)
    plot_calibration_curve(final_labels, final_probs, fold=fold)

    # ── Save validation set for XAI cell ──────────────────────────────────────
    val_df.to_csv(os.path.join(CFG.OUTPUT_DIR, "val_df.csv"), index=False)
    print(f"  val_df.csv saved → load in XAI cell")

    return model, final_auc, final_f1, final_probs, final_labels


# ─── 12. Main ─────────────────────────────────────────────────────────────────
def main():
    print("\n" + "═" * 65)
    print("  ISIC-DICM-17K  |  Transformer Fusion  v3  (XAI-Ready)")
    print("  Swin V2  ×  TabTransformer  +  Cross-Attention")
    print("═" * 65)

    df = pd.read_csv(CFG.CSV_PATH)

    def extract_id(filename):
        match = re.search(r'(\d{7})', str(filename))
        return f"ISIC_{match.group(1)}.jpg" if match else filename

    print(f"Normalizing {len(df)} image names...")
    df[CFG.IMG_COL] = df[CFG.IMG_COL].apply(extract_id)

    meta_cols = [c for c in df.columns if c not in CFG.DROP_COLS]
    validate_features(df, meta_cols)

    if CFG.USE_KFOLD:
        print(f"\nRunning {CFG.N_FOLDS}-Fold Stratified CV")
        skf        = StratifiedKFold(n_splits=CFG.N_FOLDS, shuffle=True,
                                     random_state=CFG.SEED)
        fold_aucs, fold_f1s = [], []
        oof_probs  = np.zeros(len(df))
        oof_labels = np.zeros(len(df))

        for fold, (tr_idx, vl_idx) in enumerate(
                skf.split(df, df[CFG.TARGET_COL]), 1):
            model, auc, f1, probs, labels = run_training(
                df.iloc[tr_idx], df.iloc[vl_idx], meta_cols, fold=fold
            )
            fold_aucs.append(auc)
            fold_f1s.append(f1)
            oof_probs[vl_idx]  = probs
            oof_labels[vl_idx] = labels

        oof_auc = roc_auc_score(oof_labels, oof_probs)
        print(f"\n{'═'*65}")
        print(f"  K-FOLD SUMMARY")
        print(f"  Per-fold AUC  : {[f'{a:.4f}' for a in fold_aucs]}")
        print(f"  Mean AUC      : {np.mean(fold_aucs):.4f} ± {np.std(fold_aucs):.4f}")
        print(f"  OOF AUC       : {oof_auc:.4f}")
        print(f"  Mean F1 macro : {np.mean(fold_f1s):.4f}")
        print(f"{'═'*65}")

        oof_df = pd.DataFrame({"label": oof_labels, "prob_melanoma": oof_probs})
        oof_df.to_csv(
            os.path.join(CFG.OUTPUT_DIR, "oof_predictions.csv"), index=False
        )
        print("OOF predictions saved.")

    else:
        train_df, val_df = train_test_split(
            df, test_size=0.2,
            stratify=df[CFG.TARGET_COL],
            random_state=CFG.SEED
        )
        run_training(train_df, val_df, meta_cols)

    print(f"\nAll outputs saved to: {CFG.OUTPUT_DIR}")


# ─── 13. Single-sample Inference ──────────────────────────────────────────────
def predict(image_path, metadata_series, checkpoint_path):
    """
    Classify one preprocessed image + one CSV row.
    Returns prediction, probabilities, and cross-attention weights for XAI.
    """
    # 1. Load Checkpoint
    ckpt = torch.load(checkpoint_path, map_location=CFG.DEVICE, weights_only=False)
    meta_cols = list(ckpt["meta_cols"])
    threshold = ckpt.get("threshold", 0.5)

    # 2. Rebuild Model
    model = TransformerFusionModel(meta_dim=ckpt["meta_dim"]).to(CFG.DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()

    # 3. Normalize Image Path (Regex Fix)
    # Extracts the 7-digit ID from the path/filename to match folder convention
    raw_filename = os.path.basename(image_path)
    match = re.search(r'(\d{7})', raw_filename)
    if match:
        clean_filename = f"ISIC_{match.group(1)}.jpg"
        # Join with the standard images directory from CFG
        image_path = os.path.join(CFG.IMAGES_DIR, clean_filename)
    
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"❌ Could not find processed image at: {image_path}")

    # 4. Prepare Tensors
    tf = transforms.Compose([
        transforms.Resize((CFG.IMG_SIZE, CFG.IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize(CFG.MEAN, CFG.STD),
    ])
    
    image = Image.open(image_path).convert("RGB")
    img_t = tf(image).unsqueeze(0).to(CFG.DEVICE)
    
    # Ensure only numeric values are converted to tensor
    meta_vals = metadata_series[meta_cols].values.astype(np.float32)
    meta_t = torch.tensor(meta_vals, dtype=torch.float32).unsqueeze(0).to(CFG.DEVICE)

    # 5. Inference
    with torch.no_grad():
        logits = model(img_t, meta_t)
        probs = torch.softmax(logits, dim=1)[0].cpu().numpy()
        pred = int(probs[1] >= threshold)

    # 6. Extract XAI weights
    attn_img, attn_meta = model.get_attention_weights()

    print(f"Prediction  : {'Melanoma' if pred else 'Non-Melanoma'}")
    print(f"P(Melanoma) : {probs[1]:.4f}  (threshold = {threshold:.3f})")
    
    # Visualization
    plot_cross_attention_heatmap(attn_img, meta_cols, sample_idx=0)

    return pred, probs, attn_img, attn_meta


if __name__ == "__main__":
    main()
```

```python
"""
╔══════════════════════════════════════════════════════════════════════════════╗
║          XAI CELL  —  Run AFTER the training cell has finished              ║
║                                                                              ║
║  Methods implemented:                                                        ║
║    1.  GradCAM++           — spatial heatmap on Swin last-stage feature map ║
║    2.  SHAP (DeepExpl.)    — tabular metadata feature importance            ║
║    3.  Integrated Gradients — multimodal (image + metadata) attribution     ║
║    4.  Attention Rollout   — propagated attention across all Transformer     ║
║                               layers (TabTransformer + Cross-Attention)     ║
║                                                                              ║
║  Requirements (install once):                                                ║
║      !pip install -q timm shap captum                                        ║
║                                                                              ║
║  This cell imports the model classes from the training script so it must    ║
║  be executed IN THE SAME KERNEL SESSION, or you must re-run the training    ║
║  imports at the top.                                                         ║
╚══════════════════════════════════════════════════════════════════════════════╝
"""

# ─── XAI-0: Ensure dependencies ───────────────────────────────────────────────
# Uncomment to install if needed:
# !pip install -q timm shap captum

import os
import re
import gc
import math
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors
from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import transforms

# ── Third-party XAI libraries ─────────────────────────────────────────────────
try:
    import shap
    print("✅ shap loaded")
except ImportError:
    raise ImportError("Run:  !pip install shap")

try:
    from captum.attr import IntegratedGradients, NoiseTunnel
    print("✅ captum loaded")
except ImportError:
    raise ImportError("Run:  !pip install captum")

# ── Import model classes from training script ──────────────────────────────────
# These are available because we are in the same Kaggle/Colab kernel session.
# If you run this in a fresh kernel, paste or import the class definitions.


# ──────────────────────────────────────────────────────────────────────────────
# XAI-1: CONFIGURATION
# ──────────────────────────────────────────────────────────────────────────────
class XAI_CFG:
    # Checkpoint path produced by the training cell
    CKPT_PATH   = os.path.join(CFG.OUTPUT_DIR, "best_model.pth")

    # Validation CSV saved by run_training()
    VAL_CSV     = os.path.join(CFG.OUTPUT_DIR, "val_df.csv")

    # How many samples to analyse (set low to save VRAM)
    N_SAMPLES   = 8

    # SHAP background set size (smaller = faster, noisier SHAP values)
    SHAP_BG     = 100
    SHAP_EXPLAIN  = 50
    SHAP_NSAMPLES = 200

    # Integrated Gradients steps (50 is standard; increase for smoother)
    IG_STEPS    = 50

    # Output dir (same as training)
    OUT         = CFG.OUTPUT_DIR

    DEVICE      = CFG.DEVICE


os.makedirs(XAI_CFG.OUT, exist_ok=True)


# ──────────────────────────────────────────────────────────────────────────────
# XAI-2: HELPERS
# ──────────────────────────────────────────────────────────────────────────────
def _denorm(img_tensor):
    """
    Denormalize a (C, H, W) tensor → numpy (H, W, 3) uint8, clipped [0, 255].
    Uses ImageNet mean/std stored in CFG.
    """
    mean = torch.tensor(CFG.MEAN, dtype=torch.float32).view(3, 1, 1)
    std  = torch.tensor(CFG.STD,  dtype=torch.float32).view(3, 1, 1)
    img  = img_tensor.cpu().float() * std + mean
    img  = img.permute(1, 2, 0).numpy()
    return np.clip(img * 255, 0, 255).astype(np.uint8)


def _overlay_heatmap(raw_np, heatmap_np, alpha=0.5, cmap="jet"):
    """
    Blend a (H, W) float heatmap [0,1] onto a (H, W, 3) uint8 image.
    Returns (H, W, 3) uint8 blend.
    """
    h, w = raw_np.shape[:2]
    heatmap_resized = np.array(
        Image.fromarray(
            (heatmap_np / (heatmap_np.max() + 1e-8) * 255).astype(np.uint8)
        ).resize((w, h), Image.BILINEAR)
    ).astype(np.float32) / 255.0

    colormap = plt.get_cmap(cmap)
    colored  = (colormap(heatmap_resized)[:, :, :3] * 255).astype(np.uint8)
    blended  = (raw_np.astype(np.float32) * (1 - alpha) +
                colored.astype(np.float32) * alpha).astype(np.uint8)
    return blended


import glob

def load_model_and_data():
    """
    Load the best available checkpoint + validation dataframe.
    Automatically detects fold-based or single-run naming conventions.
    """
    # 1. Auto-detect checkpoint file (finds best_model.pth or best_fold1.pth)
    ckpt_pattern = os.path.join(XAI_CFG.OUT, "best_*.pth")
    ckpt_files = sorted(glob.glob(ckpt_pattern))
    
    if not ckpt_files:
        raise FileNotFoundError(f"❌ No checkpoint found in {XAI_CFG.OUT}. "
                                "Ensure Cell 1 finished training successfully.")
    
    # Use the first/best one found
    actual_ckpt_path = ckpt_files[0]
    print(f"\nLoading checkpoint: {actual_ckpt_path}")
    
    ckpt = torch.load(actual_ckpt_path, map_location=XAI_CFG.DEVICE, weights_only=False)
    meta_cols = list(ckpt["meta_cols"])
    threshold = ckpt.get("threshold", 0.5)
    meta_dim  = ckpt["meta_dim"]

    # 2. Rebuild Model
    model = TransformerFusionModel(meta_dim=meta_dim).to(XAI_CFG.DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    
    print(f"  ✅ Loaded epoch {ckpt.get('epoch', 'N/A')} | "
          f"val AUC={ckpt.get('val_auc_raw', 0):.4f} | threshold={threshold:.3f}")

    # 3. Load and Normalize Validation Data
    if not os.path.exists(XAI_CFG.VAL_CSV):
        raise FileNotFoundError(f"❌ val_df.csv not found at {XAI_CFG.VAL_CSV}")
        
    val_df = pd.read_csv(XAI_CFG.VAL_CSV)
    
    # ── Safety Fix: Normalize image names in the validation DF ──
    def extract_id(filename):
        match = re.search(r'(\d{7})', str(filename))
        return f"ISIC_{match.group(1)}.jpg" if match else filename
    
    val_df[CFG.IMG_COL] = val_df[CFG.IMG_COL].apply(extract_id)
    # ───────────────────────────────────────────────────────────
    
    val_ds = ISICDataset(val_df, CFG.IMAGES_DIR, augment=False)

    return model, val_ds, meta_cols, threshold, val_df


# ──────────────────────────────────────────────────────────────────────────────
# XAI-3: GRAD-CAM++
# ──────────────────────────────────────────────────────────────────────────────
def run_gradcam_plusplus(model, val_ds, meta_cols, n_samples=None,
                          class_idx=1):
    """
    GradCAM++ on the last Swin V2 stage feature map.

    Algorithm:
      1. Forward pass → store last-stage feature map A (B, H, W, C).
      2. Backward pass (w.r.t. class logit) → store gradients dL/dA.
      3. Compute GradCAM++ alpha weights using second-order gradient
         approximation (Chattopadhyay et al. 2018).
      4. Weighted sum → spatial heatmap → overlay on image.

    class_idx=1  →  Melanoma class.
    """
    n_samples = n_samples or XAI_CFG.N_SAMPLES
    print(f"\n{'='*60}")
    print(f"  GradCAM++ (Melanoma class, {n_samples} samples)")
    print(f"{'='*60}")

    # Register hook that saves gradient at last Swin stage feature map
    model.img_encoder.register_gradcam_hook()
    model.train()   # need grad in batchnorm/dropouts for GradCAM

    fig, axes = plt.subplots(n_samples, 3,
                              figsize=(13, 4 * n_samples))
    if n_samples == 1:
        axes = axes[np.newaxis, :]

    for i in range(n_samples):
        img_t, meta_t, label = val_ds[i]
        img_batch  = img_t.unsqueeze(0).to(XAI_CFG.DEVICE)
        meta_batch = meta_t.unsqueeze(0).to(XAI_CFG.DEVICE)

        # Enable grad for image
        img_batch.requires_grad_(True)

        # Forward
        model.zero_grad()
        logits = model(img_batch, meta_batch)
        score  = logits[0, class_idx]

        # Backward
        score.backward()

        # ── GradCAM++ computation ──────────────────────────────────────────
        # A: feature map stored in encoder   (1, H, W, C)  or  (1, N, C)
        A    = model.img_encoder.feature_map
        dA   = model.img_encoder.grad_map     # same shape

        if A is None or dA is None:
            print(f"  Sample {i}: feature map not captured — skipping")
            continue

        # Flatten spatial dims if necessary
        if A.dim() == 4:
            # (1, H, W, C) → (1, C, H, W) for standard GradCAM formulation
            A  = A.permute(0, 3, 1, 2)        # (1, C, H, W)
            dA = dA.permute(0, 3, 1, 2)
        elif A.dim() == 3:
            # (1, N, C) — reshape to pseudo-square (1, C, sqrt_N, sqrt_N)
            N = A.shape[1]
            g = int(math.isqrt(N))
            A  = A[:, :g*g, :].permute(0, 2, 1).view(1, -1, g, g)
            dA = dA[:, :g*g, :].permute(0, 2, 1).view(1, -1, g, g)

        A  = A.detach().float()
        dA = dA.detach().float()

        # GradCAM++ alpha:
        #   alpha_k = (d^2 L/dA_k^2) / (2*(d^2 L/dA_k^2) + sum_ij(A_k * d^3/...))
        # Practical approximation (ReLU-guided, Selvaraju et al. simplified):
        relu_grad = F.relu(dA)                          # (1, C, H, W)
        # Numerator: global average of squared gradients
        numer     = relu_grad.pow(2)
        # Denominator: 2 * squared + A * cubed (channel-wise global sum)
        denom     = 2.0 * relu_grad.pow(2) + \
                    (A * relu_grad.pow(3)).sum(dim=[2, 3], keepdim=True)
        alpha     = torch.where(denom != 0, numer / (denom + 1e-8),
                                torch.zeros_like(numer))     # (1, C, H, W)

        # Weighted combination
        weights   = (alpha * relu_grad).sum(dim=[2, 3])      # (1, C)
        cam       = (weights[:, :, None, None] * A).sum(dim=1, keepdim=True)
        cam       = F.relu(cam)                              # (1, 1, H, W)
        cam_np    = cam[0, 0].cpu().numpy()

        raw_np    = _denorm(img_t)
        overlay   = _overlay_heatmap(raw_np, cam_np, alpha=0.5)

        lbl_str = "Melanoma" if label.item() == 1 else "Non-Melanoma"
        axes[i, 0].imshow(raw_np);  axes[i, 0].set_title(f"Original [{lbl_str}]")
        axes[i, 1].imshow(cam_np, cmap="jet"); axes[i, 1].set_title("CAM heatmap")
        axes[i, 2].imshow(overlay); axes[i, 2].set_title("GradCAM++ overlay")
        for ax in axes[i]:
            ax.axis("off")

    plt.suptitle("GradCAM++ — Swin V2 last-stage spatial attribution",
                 fontsize=14, fontweight="bold")
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_gradcam_plusplus.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    model.img_encoder.remove_gradcam_hook()
    model.eval()
    if XAI_CFG.DEVICE == "cuda":
        torch.cuda.empty_cache(); gc.collect()


# ──────────────────────────────────────────────────────────────────────────────
# XAI-4: SHAP (DeepExplainer on tabular metadata branch)
# ──────────────────────────────────────────────────────────────────────────────
def run_shap_tabular(model, val_ds, meta_cols, n_samples=None):
    """
    SHAP DeepExplainer on the metadata branch.

    Strategy:
      We wrap the model in a function that takes only the metadata tensor
      and fixes the image tensor to a neutral (zero) background.
      DeepExplainer computes Shapley values for each of the 23 features.

    Outputs:
      - Bar chart: mean |SHAP| per feature across all val samples
      - Beeswarm plot: SHAP value distribution per feature
    """
    n_samples = n_samples or XAI_CFG.N_SAMPLES
    print(f"\n{'='*60}")
    print(f"  SHAP DeepExplainer — tabular metadata ({n_samples} samples)")
    print(f"{'='*60}")

    # ── Collect background and test metadata tensors ───────────────────────
    all_metas  = []
    all_labels = []
    for i in range(min(len(val_ds), XAI_CFG.SHAP_BG + n_samples)):
        _, meta_t, lbl = val_ds[i]
        all_metas.append(meta_t.numpy())
        all_labels.append(lbl.item())

    all_metas  = np.array(all_metas, dtype=np.float32)
    bg_metas   = torch.tensor(all_metas[:XAI_CFG.SHAP_BG]).to(XAI_CFG.DEVICE)
    test_metas = torch.tensor(all_metas[XAI_CFG.SHAP_BG:
                                         XAI_CFG.SHAP_BG + n_samples]).to(XAI_CFG.DEVICE)

    # ── Model wrapper: metadata-only, zero image ───────────────────────────
    zero_img = torch.zeros(
        1, 3, CFG.IMG_SIZE, CFG.IMG_SIZE,
        device=XAI_CFG.DEVICE
    )

    class MetaOnlyWrapper(nn.Module):
        """
        Feeds a fixed zero image alongside the given metadata.
        Output: (B, 2) logits — SHAP targets class=1 (Melanoma).
        """
        def __init__(self, fusion_model, fixed_img):
            super().__init__()
            self.model     = fusion_model
            self.fixed_img = fixed_img

        def forward(self, meta):
            B   = meta.size(0)
            img = self.fixed_img.expand(B, -1, -1, -1)
            return self.model(img, meta)

    wrapper = MetaOnlyWrapper(model, zero_img).to(XAI_CFG.DEVICE)
    wrapper.eval()

    print(f"  Background size : {bg_metas.shape[0]}")
    print(f"  Test size       : {test_metas.shape[0]}")

    # ── DeepExplainer ─────────────────────────────────────────────────────
    explainer   = shap.DeepExplainer(wrapper, bg_metas)
    shap_values = explainer.shap_values(test_metas, nsamples=XAI_CFG.SHAP_NSAMPLES)
    # shap_values: list of [n_samples, n_features] per class
    # We want class-1 (Melanoma) values
    if isinstance(shap_values, list):
        shap_mel = np.array(shap_values[1])     # (n_samples, n_features)
    else:
        # Single output → take as is
        shap_mel = np.array(shap_values)

    test_metas_np = test_metas.cpu().numpy()

    # ── Plot 1: Mean |SHAP| bar chart ─────────────────────────────────────
    mean_abs = np.abs(shap_mel).mean(axis=0)    # (n_features,)
    order    = np.argsort(mean_abs)[::-1]

    fig, ax = plt.subplots(figsize=(14, 5))
    colors  = plt.cm.RdYlGn_r(mean_abs[order] / (mean_abs.max() + 1e-8))
    ax.barh(
        [meta_cols[j] for j in order],
        mean_abs[order],
        color=colors
    )
    ax.set_xlabel("Mean |SHAP value|  (Melanoma class)")
    ax.set_title("SHAP Feature Importance — Clinical Metadata\n"
                 "Higher = more influence on Melanoma prediction")
    ax.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_shap_bar.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    # ── Plot 2: Beeswarm (dot plot, SHAP style) ────────────────────────────
    fig, ax = plt.subplots(figsize=(14, max(6, len(meta_cols) // 2)))
    norm    = mcolors.Normalize(
        vmin=test_metas_np.min(), vmax=test_metas_np.max()
    )
    smap    = cm.ScalarMappable(norm=norm, cmap="coolwarm")

    for rank, j in enumerate(order[::-1]):   # bottom = least important
        y_pos  = rank
        x_vals = shap_mel[:, j]
        c_vals = test_metas_np[:, j]
        jitter = np.random.uniform(-0.2, 0.2, size=len(x_vals))
        ax.scatter(x_vals, y_pos + jitter,
                   c=smap.to_rgba(c_vals), s=30, alpha=0.8, zorder=3)

    ax.set_yticks(range(len(order)))
    ax.set_yticklabels([meta_cols[j] for j in order[::-1]], fontsize=8)
    ax.axvline(0, color="black", lw=0.8)
    ax.set_xlabel("SHAP value  (positive → Melanoma)")
    ax.set_title("SHAP Beeswarm — Clinical Metadata\n"
                 "Color = feature value (red=high, blue=low)")
    plt.colorbar(smap, ax=ax, label="Feature value", shrink=0.5)
    ax.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_shap_beeswarm.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    # ── Plot 3: Per-sample heatmap ─────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(14, max(4, n_samples // 2)))
    sns_data = pd.DataFrame(
        shap_mel, columns=meta_cols
    )
    import seaborn as sns
    sns.heatmap(
        sns_data, cmap="RdBu_r", center=0,
        xticklabels=True, yticklabels=[f"S{i}" for i in range(n_samples)],
        linewidths=0.3, ax=ax
    )
    ax.set_title("SHAP values per sample × feature  (Melanoma class)\n"
                 "Red = pushes toward Melanoma, Blue = away")
    plt.xticks(rotation=45, ha="right", fontsize=7)
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_shap_heatmap.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    return shap_mel, meta_cols


# ──────────────────────────────────────────────────────────────────────────────
# XAI-5: INTEGRATED GRADIENTS — MULTIMODAL
# ──────────────────────────────────────────────────────────────────────────────
def run_integrated_gradients(model, val_ds, meta_cols, n_samples=None):
    """
    Integrated Gradients (Sundararajan et al. 2017) for both modalities.

    Image IG:
      Baseline = black image (zeros in normalised space mapped to mean).
      Attribution = pixel-level importance map, visualised as abs-sum across
      channels and overlaid on the original image.

    Metadata IG:
      Baseline = zero vector (all features absent).
      Attribution = per-feature signed importance; visualised as bar chart
      alongside the feature values.

    Uses captum.attr.IntegratedGradients.
    """
    n_samples = n_samples or XAI_CFG.N_SAMPLES
    print(f"\n{'='*60}")
    print(f"  Integrated Gradients — multimodal ({n_samples} samples)")
    print(f"{'='*60}")

    # ── Wrapper: both modalities as inputs ────────────────────────────────
    class DualInputWrapper(nn.Module):
        """
        captum requires all attribution inputs to be positional arguments.
        This wrapper passes (image, meta) through the fusion model and
        returns the Melanoma logit (scalar per sample).
        """
        def __init__(self, fusion_model):
            super().__init__()
            self.model = fusion_model

        def forward(self, image, meta):
            logits = self.model(image, meta)
            # Return Melanoma class logit (index 1)
            return logits[:, 1:2]   # keep (B, 1) shape for captum

    wrapper = DualInputWrapper(model).to(XAI_CFG.DEVICE)
    wrapper.eval()

    ig = IntegratedGradients(wrapper)

    # Baselines
    img_baseline  = torch.zeros(1, 3, CFG.IMG_SIZE, CFG.IMG_SIZE,
                                device=XAI_CFG.DEVICE)
    meta_baseline = torch.zeros(1, len(meta_cols), device=XAI_CFG.DEVICE)

    fig_rows = n_samples
    fig, axes = plt.subplots(fig_rows, 4,
                              figsize=(18, 4.5 * fig_rows))
    if fig_rows == 1:
        axes = axes[np.newaxis, :]

    for i in range(n_samples):
        img_t, meta_t, label = val_ds[i]
        img_input  = img_t.unsqueeze(0).to(XAI_CFG.DEVICE)
        meta_input = meta_t.unsqueeze(0).to(XAI_CFG.DEVICE)

        # IG attribution
        attr_img, attr_meta = ig.attribute(
            inputs=(img_input, meta_input),
            baselines=(img_baseline, meta_baseline),
            n_steps=XAI_CFG.IG_STEPS,
            return_convergence_delta=False
        )
        # attr_img  : (1, 3, H, W)
        # attr_meta : (1, n_features)

        # ── Image attribution ──────────────────────────────────────────────
        attr_img_np = attr_img[0].cpu().detach().float().numpy()   # (3, H, W)
        # Absolute sum over channels → importance map
        importance_map = np.abs(attr_img_np).sum(axis=0)           # (H, W)
        # Positive / negative channels for signed visualisation
        pos_map = np.maximum(attr_img_np, 0).sum(axis=0)
        neg_map = np.maximum(-attr_img_np, 0).sum(axis=0)

        raw_np   = _denorm(img_t)
        overlay  = _overlay_heatmap(raw_np, importance_map, alpha=0.5, cmap="hot")

        lbl_str  = "Melanoma" if label.item() == 1 else "Non-Melanoma"
        axes[i, 0].imshow(raw_np)
        axes[i, 0].set_title(f"Original [{lbl_str}]")
        axes[i, 1].imshow(importance_map, cmap="hot")
        axes[i, 1].set_title("IG: |attribution| map")
        axes[i, 2].imshow(overlay)
        axes[i, 2].set_title("IG overlay on image")

        # ── Metadata attribution ───────────────────────────────────────────
        attr_meta_np = attr_meta[0].cpu().detach().float().numpy()  # (n_features,)
        meta_vals_np = meta_t.numpy()

        order     = np.argsort(np.abs(attr_meta_np))[::-1][:15]    # top-15
        bar_cols  = ["#d73027" if v >= 0 else "#4575b4"
                     for v in attr_meta_np[order]]

        axes[i, 3].barh(
            [meta_cols[j] for j in order[::-1]],
            attr_meta_np[order[::-1]],
            color=bar_cols[::-1]
        )
        axes[i, 3].axvline(0, color="black", lw=0.6)
        axes[i, 3].set_title("IG: metadata attribution (top-15)")
        axes[i, 3].set_xlabel("Attribution (pos=Melanoma)")
        axes[i, 3].tick_params(axis="y", labelsize=7)
        axes[i, 3].grid(axis="x", alpha=0.3)

        for ax in axes[i, :3]:
            ax.axis("off")

    plt.suptitle(
        "Integrated Gradients — Multimodal Attribution\n"
        "Image: pixel importance  |  Right: clinical feature importance",
        fontsize=13, fontweight="bold"
    )
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_integrated_gradients.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    if XAI_CFG.DEVICE == "cuda":
        torch.cuda.empty_cache(); gc.collect()


# ──────────────────────────────────────────────────────────────────────────────
# XAI-6: ATTENTION ROLLOUT
# ──────────────────────────────────────────────────────────────────────────────
def _rollout_single(attn_list, add_residual=True, head_fusion="mean"):
    """
    Attention Rollout (Abnar & Zuidema 2020).

    attn_list : list of (1, heads, seq, seq) tensors (one per layer).
    Returns   : (seq, seq) rolled-out attention matrix.

    Algorithm:
      A_hat = A + I   (add residual/skip connection)
      A_hat = A_hat / A_hat.sum(dim=-1, keepdim=True)  (re-normalise)
      Roll  = A_hat_layer1 @ A_hat_layer2 @ ... @ A_hat_layerN
    """
    # ── Fuse heads ────────────────────────────────────────────────────────
    if head_fusion == "mean":
        mats = [a[0].mean(dim=0) for a in attn_list]     # each (seq, seq)
    elif head_fusion == "max":
        mats = [a[0].max(dim=0).values for a in attn_list]
    else:
        mats = [a[0].mean(dim=0) for a in attn_list]

    result = torch.eye(mats[0].shape[0], device=mats[0].device)
    for mat in mats:
        if add_residual:
            mat = mat + torch.eye(mat.shape[0], device=mat.device)
        mat = mat / (mat.sum(dim=-1, keepdim=True) + 1e-8)
        result = mat @ result

    return result.cpu().numpy()   # (seq, seq)


def run_attention_rollout(model, val_ds, meta_cols, n_samples=None):
    """
    Attention Rollout across:
      A. TabTransformer self-attention layers → CLS attention over 23 features
      B. Cross-Attention layers (img→meta, meta→img) → bridging attention

    Outputs:
      - Feature importance bar chart from TabTransformer rollout
      - Cross-attention rollout heatmap (image patches × clinical features)
      - Per-sample overlay on the original image for meta→img rollout
    """
    n_samples = n_samples or XAI_CFG.N_SAMPLES
    print(f"\n{'='*60}")
    print(f"  Attention Rollout ({n_samples} samples)")
    print(f"{'='*60}")

    # ── Collect attention weights from forward passes ──────────────────────
    tab_rollouts   = []    # (n_samples, seq_meta, seq_meta)
    ca_i2m_rollouts= []   # (n_samples, seq_img, seq_meta)  image→meta
    ca_m2i_rollouts= []   # (n_samples, seq_meta, seq_img)  meta→image

    model.eval()
    with torch.no_grad():
        for i in range(n_samples):
            img_t, meta_t, label = val_ds[i]
            img_b  = img_t.unsqueeze(0).to(XAI_CFG.DEVICE)
            meta_b = meta_t.unsqueeze(0).to(XAI_CFG.DEVICE)

            _ = model(img_b, meta_b)

            # ── Tab rollout ────────────────────────────────────────────────
            tab_attn = model.get_tab_attn_weights()   # list (B,heads,seq,seq)
            tab_roll = _rollout_single(tab_attn)       # (seq_meta, seq_meta)
            tab_rollouts.append(tab_roll)

            # ── Cross-attention rollout ────────────────────────────────────
            # get_all_ca_attention_weights returns list of (attn_i2m, attn_m2i)
            ca_layers = model.get_all_ca_attention_weights()
            # Separate lists for i2m and m2i
            i2m_list = [pair[0].unsqueeze(0) if pair[0].dim() == 3
                        else pair[0][:1] for pair in ca_layers]
            m2i_list = [pair[1].unsqueeze(0) if pair[1].dim() == 3
                        else pair[1][:1] for pair in ca_layers]

            # For cross-modal rollout we just take the mean over CA layers
            # (shapes differ between i2m and m2i so we don't chain them)
            ca_i2m_mean = torch.stack(
                [p[0].unsqueeze(0).mean(dim=1) for p in ca_layers], dim=0
            ).mean(dim=0)[0].cpu().numpy()    # (seq_img, seq_meta)

            ca_m2i_mean = torch.stack(
                [p[1].unsqueeze(0).mean(dim=1) for p in ca_layers], dim=0
            ).mean(dim=0)[0].cpu().numpy()    # (seq_meta, seq_img)

            ca_i2m_rollouts.append(ca_i2m_mean)
            ca_m2i_rollouts.append(ca_m2i_mean)

    # ═══════════════════════════════════════════════════════════════════════
    # Plot A — TabTransformer Rollout: CLS → feature importance
    # ═══════════════════════════════════════════════════════════════════════
    print("\n  Plotting TabTransformer rollout (CLS → feature attention)...")

    # Average rollout over all samples; row 0 = CLS token, cols 1: = features
    avg_tab_roll = np.mean(tab_rollouts, axis=0)    # (seq_meta, seq_meta)
    cls_attn     = avg_tab_roll[0, 1:]              # (n_features,)
    n_feat       = min(len(meta_cols), len(cls_attn))
    cls_attn     = cls_attn[:n_feat]
    order        = np.argsort(cls_attn)[::-1]

    fig, ax = plt.subplots(figsize=(14, 5))
    colors  = plt.cm.YlOrRd(cls_attn[order] / (cls_attn.max() + 1e-8))
    ax.bar([meta_cols[j] for j in order], cls_attn[order], color=colors)
    ax.set_xticklabels([meta_cols[j] for j in order],
                       rotation=45, ha="right", fontsize=8)
    ax.set_ylabel("Rolled-out CLS → feature attention")
    ax.set_title(
        "Attention Rollout — TabTransformer\n"
        "How much does each clinical feature contribute through all self-attention layers?"
    )
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_rollout_tab_cls.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    # ═══════════════════════════════════════════════════════════════════════
    # Plot B — Cross-Attention Rollout heatmaps (avg over samples)
    # ═══════════════════════════════════════════════════════════════════════
    print("  Plotting cross-attention rollout heatmaps...")

    avg_i2m = np.mean(ca_i2m_rollouts, axis=0)  # (seq_img, seq_meta)
    avg_m2i = np.mean(ca_m2i_rollouts, axis=0)  # (seq_meta, seq_img)

    fig, axes = plt.subplots(1, 2, figsize=(18, 6))

    # i2m: rows=image-CLS+patches, cols=meta-CLS+features
    # Show only first row (image CLS) → attention to each meta token
    i2m_cls_row = avg_i2m[0, 1:]    # img CLS → meta features
    n_i2m       = min(len(meta_cols), len(i2m_cls_row))
    axes[0].bar(range(n_i2m), i2m_cls_row[:n_i2m],
                color=plt.cm.Purples(i2m_cls_row[:n_i2m] /
                                     (i2m_cls_row[:n_i2m].max() + 1e-8)))
    axes[0].set_xticks(range(n_i2m))
    axes[0].set_xticklabels(meta_cols[:n_i2m], rotation=45, ha="right",
                             fontsize=7)
    axes[0].set_title("CA Rollout: Image CLS → Clinical features\n"
                      "(cross-modal, averaged over CA layers)")
    axes[0].set_ylabel("Attention weight")
    axes[0].grid(axis="y", alpha=0.3)

    # m2i: meta CLS → image patches
    m2i_cls_row  = avg_m2i[0, 1:]   # meta CLS → image patches
    n_patches    = len(m2i_cls_row)
    grid_size    = int(np.sqrt(n_patches))
    if grid_size * grid_size != n_patches:
        # Pad to square
        pad = np.zeros(grid_size * grid_size)
        pad[:n_patches] = m2i_cls_row
        m2i_cls_row = pad
    spatial = m2i_cls_row.reshape(grid_size, grid_size)
    im = axes[1].imshow(spatial, cmap="hot")
    axes[1].set_title("CA Rollout: Clinical CLS → Image patches\n"
                      "(spatial attention, averaged over CA layers)")
    axes[1].axis("off")
    plt.colorbar(im, ax=axes[1], fraction=0.046)

    plt.suptitle("Cross-Attention Rollout — Bidirectional Modal Interaction",
                 fontsize=13, fontweight="bold")
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_rollout_crossattn.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    # ═══════════════════════════════════════════════════════════════════════
    # Plot C — Per-sample meta→image rollout overlay
    # ═══════════════════════════════════════════════════════════════════════
    print("  Plotting per-sample meta→image rollout overlays...")

    fig, axes = plt.subplots(n_samples, 2,
                              figsize=(10, 4 * n_samples))
    if n_samples == 1:
        axes = axes[np.newaxis, :]

    for i in range(n_samples):
        img_t, _, label = val_ds[i]
        raw_np = _denorm(img_t)

        m2i_row   = ca_m2i_rollouts[i][0, 1:]    # meta CLS → patches
        n_p       = len(m2i_row)
        gs        = int(np.sqrt(n_p))
        if gs * gs != n_p:
            pad = np.zeros(gs * gs)
            pad[:n_p] = m2i_row
            m2i_row = pad
        spatial   = m2i_row.reshape(gs, gs)
        overlay   = _overlay_heatmap(raw_np, spatial, alpha=0.55, cmap="plasma")

        lbl_str   = "Melanoma" if label.item() == 1 else "Non-Mel"
        axes[i, 0].imshow(raw_np); axes[i, 0].set_title(f"Original [{lbl_str}]")
        axes[i, 1].imshow(overlay); axes[i, 1].set_title("Rollout overlay")
        for ax in axes[i]:
            ax.axis("off")

    plt.suptitle("Attention Rollout: Where does clinical metadata 'look' in the image?",
                 fontsize=12, fontweight="bold")
    plt.tight_layout()
    save_path = os.path.join(XAI_CFG.OUT, "xai_rollout_overlay.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    if XAI_CFG.DEVICE == "cuda":
        torch.cuda.empty_cache(); gc.collect()


# ──────────────────────────────────────────────────────────────────────────────
# XAI-7: SUMMARY FIGURE  (all methods side-by-side for one sample)
# ──────────────────────────────────────────────────────────────────────────────
def run_summary_panel(model, val_ds, meta_cols, sample_idx=0):
    """
    Produce a single publication-quality panel combining all XAI methods
    for one selected sample.

    Layout (2 rows × 4 cols):
      Row 0: Original | GradCAM++ | IG image | Rollout overlay
      Row 1: SHAP bar | IG meta   | Rollout tab CLS | (blank / caption)
    """
    print(f"\n{'='*60}")
    print(f"  Summary Panel — sample index {sample_idx}")
    print(f"{'='*60}")

    img_t, meta_t, label = val_ds[sample_idx]
    img_b  = img_t.unsqueeze(0).to(XAI_CFG.DEVICE)
    meta_b = meta_t.unsqueeze(0).to(XAI_CFG.DEVICE)
    raw_np = _denorm(img_t)
    lbl_str = "Melanoma" if label.item() == 1 else "Non-Melanoma"

    # ── 1. GradCAM++ ──────────────────────────────────────────────────────
    model.img_encoder.register_gradcam_hook()
    model.train()
    model.zero_grad()
    logits = model(img_b, meta_b)
    logits[0, 1].backward()

    A  = model.img_encoder.feature_map
    dA = model.img_encoder.grad_map

    if A is not None and dA is not None:
        if A.dim() == 4:
            A  = A.permute(0, 3, 1, 2)
            dA = dA.permute(0, 3, 1, 2)
        elif A.dim() == 3:
            N = A.shape[1]; g = int(math.isqrt(N))
            A  = A[:, :g*g, :].permute(0, 2, 1).view(1, -1, g, g)
            dA = dA[:, :g*g, :].permute(0, 2, 1).view(1, -1, g, g)
        A = A.detach().float(); dA = dA.detach().float()
        relu_grad = F.relu(dA)
        numer     = relu_grad.pow(2)
        denom     = 2.0 * relu_grad.pow(2) + \
                    (A * relu_grad.pow(3)).sum(dim=[2, 3], keepdim=True)
        alpha     = torch.where(denom != 0, numer / (denom + 1e-8),
                                torch.zeros_like(numer))
        weights   = (alpha * relu_grad).sum(dim=[2, 3])
        cam       = F.relu((weights[:, :, None, None] * A).sum(dim=1))
        cam_np    = cam[0].cpu().numpy()
        gradcam_overlay = _overlay_heatmap(raw_np, cam_np)
    else:
        cam_np = np.zeros((8, 8))
        gradcam_overlay = raw_np.copy()

    model.img_encoder.remove_gradcam_hook()
    model.eval()
    model.zero_grad()

    # ── 2. Integrated Gradients ───────────────────────────────────────────
    class DualInputWrapper(nn.Module):
        def __init__(self, m): super().__init__(); self.model = m
        def forward(self, img, meta): return self.model(img, meta)[:, 1:2]

    wrapper = DualInputWrapper(model).to(XAI_CFG.DEVICE)
    ig      = IntegratedGradients(wrapper)
    attr_img, attr_meta = ig.attribute(
        inputs=(img_b, meta_b),
        baselines=(torch.zeros_like(img_b), torch.zeros_like(meta_b)),
        n_steps=XAI_CFG.IG_STEPS,
        return_convergence_delta=False
    )
    ig_map    = np.abs(attr_img[0].cpu().detach().float().numpy()).sum(axis=0)
    ig_overlay= _overlay_heatmap(raw_np, ig_map, alpha=0.5, cmap="hot")
    attr_meta_np = attr_meta[0].cpu().detach().float().numpy()

    # ── 3. Attention Rollout ──────────────────────────────────────────────
    with torch.no_grad():
        _ = model(img_b, meta_b)

    tab_attn = model.get_tab_attn_weights()
    tab_roll = _rollout_single(tab_attn)
    tab_cls  = tab_roll[0, 1:len(meta_cols)+1]

    ca_layers   = model.get_all_ca_attention_weights()
    m2i_mean    = torch.stack(
        [p[1].unsqueeze(0).mean(dim=1) for p in ca_layers], dim=0
    ).mean(dim=0)[0].cpu().numpy()
    m2i_patches = m2i_mean[0, 1:]
    N  = len(m2i_patches); g = int(math.isqrt(N))
    if g * g != N:
        pad = np.zeros(g * g); pad[:N] = m2i_patches; m2i_patches = pad
    spatial_roll = m2i_patches.reshape(g, g)
    rollout_overlay = _overlay_heatmap(raw_np, spatial_roll,
                                       alpha=0.55, cmap="plasma")

    # ── SHAP (quick, 1-sample) ─────────────────────────────────────────────
    bg_limit = min(XAI_CFG.SHAP_BG, len(val_ds))
    bg_metas = torch.stack([val_ds[j][1] for j in range(bg_limit)]).to(XAI_CFG.DEVICE)
    zero_img = torch.zeros(1, 3, CFG.IMG_SIZE, CFG.IMG_SIZE,
                           device=XAI_CFG.DEVICE)

    class MetaWrapper(nn.Module):
        def __init__(self, m, zi): super().__init__(); self.m = m; self.zi = zi
        def forward(self, meta):
            return self.m(self.zi.expand(meta.size(0), -1, -1, -1), meta)

    mwrap     = MetaWrapper(model, zero_img).to(XAI_CFG.DEVICE)
    mwrap.eval()
    explainer = shap.DeepExplainer(mwrap, bg_metas)
    sv        = explainer.shap_values(meta_b)
    shap_vals = np.array(sv[1] if isinstance(sv, list) else sv)[0]

    # ── LAYOUT ────────────────────────────────────────────────────────────
    fig = plt.figure(figsize=(22, 11))
    gs  = fig.add_gridspec(2, 4, hspace=0.4, wspace=0.35)

    # Row 0
    ax00 = fig.add_subplot(gs[0, 0])
    ax00.imshow(raw_np); ax00.set_title(f"Original\n[{lbl_str}]", fontweight="bold")
    ax00.axis("off")

    ax01 = fig.add_subplot(gs[0, 1])
    ax01.imshow(gradcam_overlay); ax01.set_title("GradCAM++\n(last Swin stage)")
    ax01.axis("off")

    ax02 = fig.add_subplot(gs[0, 2])
    ax02.imshow(ig_overlay); ax02.set_title("Integrated Gradients\n(image attribution)")
    ax02.axis("off")

    ax03 = fig.add_subplot(gs[0, 3])
    ax03.imshow(rollout_overlay); ax03.set_title("Attention Rollout\nmeta→image overlay")
    ax03.axis("off")

    # Row 1: bar charts
    ax10 = fig.add_subplot(gs[1, 0])
    top_shap = np.argsort(np.abs(shap_vals))[::-1][:10]
    cols_s   = ["#d73027" if shap_vals[j] >= 0 else "#4575b4" for j in top_shap]
    ax10.barh([meta_cols[j] for j in top_shap[::-1]],
              shap_vals[top_shap[::-1]], color=cols_s[::-1])
    ax10.axvline(0, color="k", lw=0.6)
    ax10.set_title("SHAP (top-10 features)\nred=Melanoma push")
    ax10.tick_params(axis="y", labelsize=7)
    ax10.grid(axis="x", alpha=0.3)

    ax11 = fig.add_subplot(gs[1, 1])
    top_ig = np.argsort(np.abs(attr_meta_np))[::-1][:10]
    cols_i  = ["#d73027" if attr_meta_np[j] >= 0 else "#4575b4" for j in top_ig]
    ax11.barh([meta_cols[j] for j in top_ig[::-1]],
              attr_meta_np[top_ig[::-1]], color=cols_i[::-1])
    ax11.axvline(0, color="k", lw=0.6)
    ax11.set_title("Integrated Gradients\n(top-10 meta features)")
    ax11.tick_params(axis="y", labelsize=7)
    ax11.grid(axis="x", alpha=0.3)

    ax12 = fig.add_subplot(gs[1, 2])
    top_roll = np.argsort(tab_cls)[::-1][:10]
    n_show   = min(10, len(meta_cols))
    ax12.barh(
        [meta_cols[j] for j in top_roll[:n_show][::-1]],
        tab_cls[top_roll[:n_show][::-1]],
        color=plt.cm.YlOrRd(
            tab_cls[top_roll[:n_show]] / (tab_cls.max() + 1e-8)
        )[::-1]
    )
    ax12.set_title("Attention Rollout\n(TabTransformer CLS, top-10)")
    ax12.tick_params(axis="y", labelsize=7)
    ax12.grid(axis="x", alpha=0.3)

    ax13 = fig.add_subplot(gs[1, 3])
    ax13.axis("off")
    caption = (
        "XAI Summary Panel\n\n"
        "GradCAM++: pixel-level spatial\n"
        "importance from gradient × activation\n\n"
        "Integrated Gradients: path-integrated\n"
        "attribution for image & metadata\n\n"
        "SHAP: Shapley-value feature importance\n"
        "for clinical metadata branch\n\n"
        "Attention Rollout: propagated attention\n"
        "across all Transformer layers"
    )
    ax13.text(0.05, 0.95, caption, transform=ax13.transAxes,
              va="top", ha="left", fontsize=9,
              bbox=dict(boxstyle="round,pad=0.5", fc="#f0f0f0", alpha=0.8))

    fig.suptitle(
        f"XAI Summary — Swin V2 × TabTransformer Multimodal Fusion\n"
        f"Sample #{sample_idx}  |  Ground truth: {lbl_str}",
        fontsize=14, fontweight="bold"
    )
    save_path = os.path.join(XAI_CFG.OUT, f"xai_summary_sample{sample_idx}.png")
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show(); plt.close()
    print(f"  Saved: {save_path}")

    if XAI_CFG.DEVICE == "cuda":
        torch.cuda.empty_cache(); gc.collect()


# ──────────────────────────────────────────────────────────────────────────────
# XAI-8: MAIN ENTRY POINT
# ──────────────────────────────────────────────────────────────────────────────
def run_all_xai():
    """
    Master function — runs all four XAI methods and the summary panel.
    Call this cell after the training cell has completed.
    """
    print("\n" + "═" * 65)
    print("  XAI CELL  —  SHAP | GradCAM++ | Integrated Gradients | Rollout")
    print("═" * 65)

    model, val_ds, meta_cols, threshold, val_df = load_model_and_data()

    # ── Method 1: GradCAM++ ───────────────────────────────────────────────
    run_gradcam_plusplus(model, val_ds, meta_cols,
                          n_samples=XAI_CFG.N_SAMPLES)

    # ── Method 2: SHAP ────────────────────────────────────────────────────
    run_shap_tabular(model, val_ds, meta_cols,
                     n_samples=XAI_CFG.N_SAMPLES)

    # ── Method 3: Integrated Gradients ───────────────────────────────────
    run_integrated_gradients(model, val_ds, meta_cols,
                              n_samples=XAI_CFG.N_SAMPLES)

    # ── Method 4: Attention Rollout ───────────────────────────────────────
    run_attention_rollout(model, val_ds, meta_cols,
                          n_samples=XAI_CFG.N_SAMPLES)

    # ── Summary Panel ─────────────────────────────────────────────────────
    run_summary_panel(model, val_ds, meta_cols, sample_idx=0)

    print(f"\n{'═'*65}")
    print(f"  All XAI outputs saved to: {XAI_CFG.OUT}")
    print(f"{'═'*65}")


# ─── Run ──────────────────────────────────────────────────────────────────────
run_all_xai()
```
